```python
# Find if a number is prime or composite
# prime number is one that will not get divided by any other number except 1 and itself
a = int(input('enter a number'))
if a == 1 or a == 0:
  print('Neither prime nor composite')
if a<0:
  print('Please enter only positive numbers')
if a == 2 or a == 3:
  print('The number is prime')
for i in range(2,a//i):
  print('Testing with', i)
  if a%i==0:
    print('It is not prime')
    break
else:
  print('Its prime')
```

```python
# map, filter and reduce
# I have the year of birth of all students in a list form. I want to find their age
# 2000, 2026 -----> 26 years old
a = [1990,1992,2000, 2001, 1998, 1999, 2002, 2003]
b = []
for i in a:
  b.append(2026-i)
b
```

```python
# map - apply a transformation on every element of a iterable (list or tuple)
result = map(lambda x: 2026-x, a)
print(list(result))
```

```python
# filter function
# how do you find a number is even or odd?
x = 5
if x%2 == 0:
  print('even')
else:
  print('odd')

y = lambda x:x%2==0
print(y(5))

y = lambda x: 'Even' if x%2==0 else 'Odd'
print(y(5))
```

```python
a = [2,3,4,5,6,7,54,61,8,4,3,5,88]
# i want to filter out even numbers
b = []
for i in a:
  if i%2==0:
    b.append(i)
print(b)
```

```python
y = filter(lambda x:x%2==0,a)
print(list(y))
```

```python
# reduce
# Cumulative addition
a = [2,3,4,5,6,7,54,61,8,4,3,5,88]
b = 0
for i in a:
  b = b+i
print(b)
```

```python
from functools import reduce
result = reduce(lambda x,y:x+y, a)
print(result)
```

```python
result = reduce(lambda x,y:x if x>y else y, a)
result
```
